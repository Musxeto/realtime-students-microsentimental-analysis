```python
def run_live_inference():
    """Run YOLOv11 inference on webcam feed."""
    
    # Open webcam
    cap = cv2.VideoCapture(WEBCAM_INDEX)
    
    if not cap.isOpened():
        print("❌ Could not open webcam")
        return
    
    print("🎥 Starting live inference... Press 'q' to quit")
    print("=" * 60)
    
    frame_count = 0
    start_time = time.time()
    
    while True:
        success, frame = cap.read()
        
        if not success:
            print("⚠️ Failed to read frame from webcam")
            break
        
        # Run inference with confidence threshold
        results = model(frame, conf=CONFIDENCE_THRESHOLD)
        
        # Draw annotations on frame
        annotated_frame = results[0].plot()
        
        # Display FPS and frame count
        frame_count += 1
        elapsed = time.time() - start_time
        fps = frame_count / elapsed if elapsed > 0 else 0
        
        # Add FPS text
        cv2.putText(
            annotated_frame,
            f"FPS: {fps:.1f} | Detections: {len(results[0].boxes)}",
            (10, 30),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.7,
            (0, 255, 0),
            2
        )
        
        # Display the frame
        cv2.imshow("Real-Time Student Behavior Detection", annotated_frame)
        
        # Check for 'q' key press to quit
        if cv2.waitKey(1) & 0xFF == ord('q'):
            print(f"\n🛑 Quit requested")
            break
    
    # Cleanup
    cap.release()
    cv2.destroyAllWindows()
    
    print("=" * 60)
    print(f"✅ Inference completed!")
    print(f"   Total frames processed: {frame_count}")
    print(f"   Average FPS: {frame_count / elapsed:.2f}")

# Run inference
run_live_inference()
```
